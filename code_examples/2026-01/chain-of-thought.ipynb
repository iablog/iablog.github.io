```python
import google.genai as genai
import os

from google.colab import userdata

GEMINI_API_KEY = userdata.get("GOOGLE_API_KEY")
```

```python
client = genai.Client(api_key=GEMINI_API_KEY)
```

```python
modelos = client.models.list()
for model in modelos:
    if "generateContent" in model.supported_actions:
        print(f"Nome do Modelo: {model.name}")
```

```python
#model = client.GenerativeModel('gemma-3-1b-it')
#model = genai.models.GenerativeModel('gemma-3-1b-it')
model = client.models.get(model='gemma-3-1b-it')
```

```python
prompt = """Um restaurante tem 23 mesas. Cada mesa pode acomodar 4 pessoas.
Se 78 pessoas chegarem, quantas mesas ficarão vazias?"""

resposta = client.models.generate_content(
    model='gemma-3-1b-it',
    contents=prompt,
)

print(resposta.text)
```
